```python
class MapModelDataset(Dataset):

    def __init__(self, map_dirs, model_dir):
        # map_dirs: a list of map directories
        self.map_dirs = map_dirs
        self.model_dir = model_dir
        # self.maps: a list of lists of map_names
        self.maps = [sorted(os.listdir(map_dir)) for map_dir in map_dirs]
        self.models = sorted(os.listdir(model_dir))

    def __len__(self):
        return len(self.maps)

    def __getitem__(self, index):
        map = None
        for i in range(len(self.map_dirs)):
            map_dir = self.map_dirs[i]
            map_name = self.maps[i][index]
            map_path = os.path.join(map_dir, map_name)
            # add a dimension for feature
            map_part = np.expand_dims(np.load(map_path), axis=0)
            # concatenate maps in different features
            if map is None:
                map = map_part
            else:
                map = np.concatenate((map, map_part), axis=0)
            # print(map.shape)

        model_path = os.path.join(self.model_dir, self.models[index])
        model = np.load(model_path)
        # model = np.expand_dims(np.load(model_path), axis=0)
        print(map.shape)
        print(model.shape)
        print(np.sum(map), np.sum(model))

        return map, model
```

```python
class MapModelDataset(Dataset):

    def __init__(self, map_dir, model_dir, model_type):
        self.map_dir = map_dir
        self.model_dir = model_dir
        self.maps = os.listdir(map_dir)
        self.model_type = model_type

    def __len__(self):
        return len(self.maps)

    def __getitem__(self, index):
        map_path = os.path.join(self.map_dir, self.maps[index])
        model_path = os.path.join(
            self.model_dir,
            self.maps[index].replace('map', "model_" + str(self.model_type)))
        map = np.expand_dims(np.load(map_path),
                             axis=0)  # add feature dimension
        # print(map.shape)
        model = np.load(model_path)
        # model = np.expand_dims(np.load(model_path), axis=0)         # add feature dimension
        # print(model.shape)
        print(map.shape)
        print(np.max(map), np.max(model))
        print(model.shape)
        return map, model
```
