```python
import os

import numpy as np
from torch.utils.data import Dataset
```

```python
map_dir = "/home/qiboxu/Database/U_NET/EMDB_PDB_for_U_Net/test_for_dirrerentiaiton_ssRNA/spliceosome_re2_3/ribosome/EMD-13112_re_2.4 - GOOD - ribosome"
maps = os.listdir(map_dir)
print(maps)
```

```python
class MapModelDataset(Dataset):

    def __init__(self, map_dir, model_dir, model_type):
        self.map_dir = map_dir
        self.model_dir = model_dir
        self.maps = os.listdir(map_dir)
        self.model_type = model_type

    def __len__(self):
        return len(self.maps)

    def __getitem__(self, index):
        map_path = os.path.join(self.map_dir, self.maps[index])
        model_path = os.path.join(
            self.model_dir,
            self.maps[index].replace('map', "model_" + str(self.model_type)))
        # add feature dimension
        map = np.expand_dims(np.load(map_path), axis=0)
        # print(map.shape)
        model = np.load(model_path)
        # model = np.expand_dims(np.load(model_path), axis=0)         # add feature dimension
        # print(model.shape)
        return map, model
```
